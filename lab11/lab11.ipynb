```python
import re
import string
import emoji
import urllib.request

taka_historia = """<p> Cześć!!! 👋 Witajcie na moim nowym blogu o sztucznej inteligencji...   Dzisiaj porozmawiamy o NLP (Natural Language Processing). </p>

Czy wiedzieliście, że aż 80% danych w firmach to dane nieustrukturyzowane??? 😲 Więcej informacji znajdziecie na stronie: https://www.przykladowastrona.pl/nlp-wstep lub pisząc na e-mail: kontakt@moj-blog-ai.com.pl.

#MachineLearning #DataScience @Kowalski_Data_Geek

W      niektórych    miejscach celowo zostawiłem  duuuuuużo spacji i tabulacji. ALBO NAPISAŁEM COŚ CAPSLOCKIEM, żebyście mieli co zmieniać na małe litery (tzw. lowercasing).
Warto usunąć z tego tekstu polskie stop-words, np.: "i", "w", "na", "oraz", "że".

Data publikacji: 12.05.2023 r., godz. 14:30. Zysk firmy wzrósł o $45,000 w Q3!
P.S. Nie zapomnijcie o usunięciu tagów HTML, np. <b>pogrubienia</b> i znaków interpunkcyjnych! 🚀"""


##zadanie1
tekst_1 = taka_historia

tekst_1 = re.sub(r'<[^>]+>', '', tekst_1)

tekst_1 = emoji.demojize(tekst_1)

tekst_1 = tekst_1.lower()

znaki_przestankowe = re.escape(string.punctuation)
tekst_1 = re.sub(r'([' + znaki_przestankowe + r'])\1+', r'\1', tekst_1)

tekst_1 = re.sub(r'\s+', ' ', tekst_1).strip()

print(tekst_1)
print("\n")


##zadanie2
tekst_2 = taka_historia

tekst_2 = re.sub(r'https?://\S+|www\.\S+', '<URL>', tekst_2)

tekst_2 = re.sub(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b', '<EMAIL>', tekst_2)

tekst_2 = re.sub(r'\$?\d+(?:[.,:]\d+)*', '<NUM>', tekst_2)

print(tekst_2)
print("\n")


##zadanie3
url = "https://raw.githubusercontent.com/bieli/stopwords/master/polish.stopwords.txt"
response = urllib.request.urlopen(url)
stopwords = response.read().decode('utf-8').splitlines()

wzorzec_stopwords = r'\b(?:' + '|'.join(re.escape(word) for word in stopwords) + r')\b'

tekst_3 = re.sub(wzorzec_stopwords, '', tekst_1, flags=re.IGNORECASE)

tekst_3 = re.sub(r'\s+', ' ', tekst_3).strip()

print(tekst_3)
```
