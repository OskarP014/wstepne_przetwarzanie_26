```python
import pandas as pd
import numpy as np
from statsmodels.stats.outliers_influence import variance_inflation_factor


#1.1
df = pd.read_csv('titanic.csv')

procent_brakow = (df.isna().sum() / len(df)) * 100
print("Procent brakujących wartości")
print(procent_brakow)
print("\n")

df['Embarked'] = df['Embarked'].fillna('U')

df['Age'] = df['Age'].fillna(df['Age'].mean())


##1.2
df_kopia1 = df.copy()

df_kopia1 = pd.get_dummies(df_kopia1, columns=['Sex', 'Embarked'], drop_first=False)

df_kopia1 = df_kopia1.astype({col: int for col in df_kopia1.columns if col.startswith(('Sex_', 'Embarked_'))})


#1.3
print("1.3")
kolumny_numeryczne1 = df_kopia1.select_dtypes(include=[np.number]).columns
kolumny_numeryczne1 = kolumny_numeryczne1.drop(['PassengerId', 'Survived'], errors='ignore')
X1 = df_kopia1[kolumny_numeryczne1]

vif_data1 = pd.DataFrame()
vif_data1["Cecha"] = X1.columns
vif_data1["VIF"] = [variance_inflation_factor(X1.values, i) for i in range(len(X1.columns))]
print(vif_data1)
print("\n")


#1.4
print("1.4")
df_kopia2 = df.copy()

df_kopia2 = pd.get_dummies(df_kopia2, columns=['Sex', 'Embarked'], drop_first=True)
df_kopia2 = df_kopia2.astype({col: int for col in df_kopia2.columns if col.startswith(('Sex_', 'Embarked_'))})


kolumny_numeryczne2 = df_kopia2.select_dtypes(include=[np.number]).columns
kolumny_numeryczne2 = kolumny_numeryczne2.drop(['PassengerId', 'Survived'], errors='ignore')
X2 = df_kopia2[kolumny_numeryczne2]

vif_data2 = pd.DataFrame()
vif_data2["Cecha"] = X2.columns
vif_data2["VIF"] = [variance_inflation_factor(X2.values, i) for i in range(len(X2.columns))]
print(vif_data2)
```
